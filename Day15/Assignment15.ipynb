```python
import sqlite3
import pandas as pd

# 1. Connect to an in-memory SQLite database
connection = sqlite3.connect(':memory:')
cursor = connection.cursor()

# 2. Create the 'Departments' table
cursor.execute('''
CREATE TABLE Departments (
    DepartmentID INTEGER PRIMARY KEY,
    DepartmentName TEXT NOT NULL,
    Location TEXT
)
''')

# 3. Create the 'Employees' table (Linked via DepartmentID)
cursor.execute('''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    FirstName TEXT NOT NULL,
    LastName TEXT NOT NULL,
    Role TEXT,
    Salary INTEGER,
    DepartmentID INTEGER,
    FOREIGN KEY (DepartmentID) REFERENCES Departments(DepartmentID)
)
''')

# 4. Insert Sample Data for Departments
departments_data = [
    (101, 'Data Science', 'New York'),
    (102, 'Engineering', 'San Francisco'),
    (103, 'Marketing', 'London')
]
cursor.executemany('INSERT INTO Departments VALUES (?,?,?)', departments_data)

# 5. Insert Sample Data for Employees
employees_data = [
    (1, 'Sahil', 'Patil', 'Data Analyst', 75000, 101),
    (2, 'Alice', 'Smith', 'Software Engineer', 95000, 102),
    (3, 'Bob', 'Johnson', 'ML Engineer', 105000, 101),
    (4, 'Charlie', 'Brown', 'SEO Specialist', 60000, 103),
    (5, 'David', 'Wilson', 'Backend Developer', 88000, 102)
]
cursor.executemany('INSERT INTO Employees VALUES (?,?,?,?,?,?)', employees_data)

# 6. Verify and Display Results
print("--- Employees Table ---")
print(pd.read_sql_query("SELECT * FROM Employees", connection))

print("\n--- Departments Table ---")
print(pd.read_sql_query("SELECT * FROM Departments", connection))

# 7. Test a Relational Query (Joining the tables)
query = """
SELECT e.FirstName, e.Role, d.DepartmentName
FROM Employees e
JOIN Departments d ON e.DepartmentID = d.DepartmentID
"""
print("\n--- Joined Results (Text-to-SQL Test) ---")
print(pd.read_sql_query(query, connection))
```
