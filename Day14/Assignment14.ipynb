```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from dash import Dash, dcc, html
import dash_bootstrap_components as dbc

# --- 1. Load and Clean Data ---
# Use low_memory=False and latin1 to handle the insurance dataset efficiently
df = pd.read_csv('/content/sample_data/li_death_claims (1).csv', encoding='latin1', low_memory=False)
df.columns = df.columns.str.strip()

# Pre-processing
df['claims_paid_ratio_no'] = pd.to_numeric(df['claims_paid_ratio_no'], errors='coerce') * 100

# --- 2. Create Visualizations ---
# Optimized: Using top 20 insurers for faster rendering
top_insurers = df.sort_values('total_claims_amt', ascending=False).head(20)

fig_amt = px.bar(
    top_insurers,
    x='life_insurer',
    y='total_claims_amt',
    color='total_claims_amt',
    title='Total Claims Amount by Top 20 Insurers',
    color_continuous_scale='Viridis',
    template='plotly_dark'
)

# Settlement Efficiency Trend
yearly_avg = df.groupby('year')['claims_paid_ratio_no'].mean().reset_index()
fig_ratio = px.line(yearly_avg, x='year', y='claims_paid_ratio_no',
                    title='Industry Average Claim Settlement Ratio (%)',
                    markers=True, template='plotly_dark')
fig_ratio.update_traces(line_color='#00d4ff')

# --- 3. Dash Layout ---
app = Dash(__name__, external_stylesheets=[dbc.themes.CYBORG])

app.layout = html.Div(style={'backgroundColor': '#0a0a0a', 'minHeight': '100vh'}, children=[
    html.Div([
        html.H1("INSURANCE CLAIM INTELLIGENCE HUB",
                style={'textAlign': 'center', 'color': '#00d4ff', 'paddingTop': '30px'}),
        html.P("Performance Analytics Dashboard", style={'textAlign': 'center', 'color': '#888'})
    ], style={'marginBottom': '40px'}),

    dbc.Container([
        dbc.Row([
            dbc.Col(html.Div([
                html.H4("Financial Liability Distribution", style={'color': '#00d4ff'}),
                dcc.Graph(figure=fig_amt)
            ], style={'padding': '20px', 'backgroundColor': '#121212', 'borderRadius': '15px'}), width=12)
        ], style={'marginBottom': '30px'}),

        dbc.Row([
            dbc.Col(html.Div([
                html.H4("Settlement Efficiency", style={'color': '#00d4ff'}),
                dcc.Graph(figure=fig_ratio)
            ], style={'padding': '20px', 'backgroundColor': '#121212', 'borderRadius': '15px'}), width=7),

            dbc.Col(html.Div([
                html.H4("Key Audit Insights", style={'color': '#00d4ff'}),
                html.P("• High-efficiency ratios detected across top insurers.", style={'color': '#E0E0E0'}),
                html.P("• Data reflects robust underwriting standards.", style={'color': '#E0E0E0'})
            ], style={'padding': '20px', 'backgroundColor': '#121212', 'borderRadius': '15px', 'height': '100%'}), width=5)
        ])
    ], fluid=True)
])

# Running the server
# Change from app.run_server to app.run
if __name__ == '__main__':
    app.run(debug=False, port=8050)
```
