```python
# 1. Unified Installation (Run this first)
!pip install -q -U google-genai
!pip install -q kaggle

import os
import json
from google.colab import userdata, files
from google import genai
from PIL import Image
import IPython.display

# 2. Setup Client (Only one client needed)
API_KEY = userdata.get('GOOGLE_API_KEY')
client = genai.Client(api_key=API_KEY)

# 3. Kaggle Setup (Unchanged logic)
print("Please upload your kaggle.json file:")
uploaded = files.upload()

!mkdir -p ~/.kaggle
!cp kaggle.json ~/.kaggle/
!chmod 600 ~/.kaggle/kaggle.json

# 4. Download and unzip
!kaggle datasets download -d saeedehkamjoo/standard-test-images
!unzip -o -q standard-test-images.zip -d my_images  # Added -o to overwrite if re-running

# 5. Get Image Paths
image_folder = "./my_images"
image_paths = [os.path.join(image_folder, f) for f in os.listdir(image_folder)
               if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
print(f"Ingested {len(image_paths)} images successfully.\n")

# 6. Optimized RAG Function
def implement_image_rag(query, image_list):
    """
    Simulates Multimodal RAG as per assignment requirements.
    """
    print(f"Executing Intelligent Search for: '{query}'\n")

    # Process the 15-20 images required by the assignment
    for path in image_list[:20]:
        try:
            img = Image.open(path)

            # The Generation step: LLM analyzes image context vs query
            prompt = f"Analyze this image. If it contains {query}, describe the visual match in one sentence. If not, output 'NO_MATCH'."

            response = client.models.generate_content(
                model="gemini-1.5-flash",
                contents=[prompt, img]
            )

            # The Retrieval verification
            if "NO_MATCH" not in response.text.upper():
                print(f"--- MATCH FOUND: {os.path.basename(path)} ---")
                IPython.display.display(img.resize((250, 250)))
                print(f"AI Generation: {response.text}\n")
                print("-" * 30)
        except Exception as e:
            continue

# 7. Execute Generation
implement_image_rag("an animal or a bird", image_paths)
```
