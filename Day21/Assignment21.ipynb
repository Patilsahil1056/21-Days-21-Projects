```python


import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, SystemMessage

# 1. Setup your Google API Key
# Get your key from https://aistudio.google.com/
os.environ["GOOGLE_API_KEY"] = "AIzaSyCUJLfCNWtU_l-qEaarvkSkSQt2TGIDXCo"

# 2. Initialize the Gemini Model (The "Brain" node in n8n)
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")


# 3. Define the Agent's Instructions (The "System Prompt")
system_instruction = SystemMessage(content="""
You are a helpful AI Agent. Your goal is to assist users with their tasks.
If the user asks a question, answer it concisely.
If the user asks for a summary, provide 3 clear bullet points.
""")

def run_agent(user_input):
    print(f"\nUser: {user_input}")
    # Combine the instructions with the specific user question
    messages = [system_instruction, HumanMessage(content=user_input)]
    response = llm.invoke(messages)
    print(f"Agent: {response.content}")

# 4. Running the Agent with multiple questions
print("--- Starting your First AI Agent ---")

# Question 1: Introduction
run_agent("Hello! Who are you and what can you do?")

# Question 2: Testing Task capabilities (Summarization)
run_agent("Can you summarize the benefits of using AI for newsletters in 3 points?")

# Question 3: Technical question
run_agent("What is the difference between an LLM and an AI Agent?")
```
