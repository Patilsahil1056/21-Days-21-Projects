```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2

# Load the model
model = YOLO('yolov8n.pt')

# Path to the image
img_path = 'https://raw.githubusercontent.com/ultralytics/ultralytics/main/ultralytics/assets/bus.jpg'

# Run detection once (NOT in a loop)
results = model(img_path)

# Plot the annotated result
annotated_img = cv2.cvtColor(results[0].plot(), cv2.COLOR_BGR2RGB)
plt.imshow(annotated_img)
plt.title("YOLOv8 Prediction")
plt.axis('off')
plt.show()
```
